```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

tickers = ['AAPL', 'TSLA', 'NVDA']
headers = {'User-Agent': 'Mozilla/5.0'}
rows = []

for t in tickers:
    url = f'https://finviz.com/quote.ashx?t={t}'
    r = requests.get(url, headers=headers)
    print(t, r.status_code)
    soup = BeautifulSoup(r.text, 'html.parser')
    table = soup.find(id='news-table')
    if table is None:
        print('  ما لقيت جدول الأخبار')
        continue
    for tr in table.find_all('tr'):
        a = tr.find('a')
        if a:
            rows.append({'ticker': t, 'time': tr.td.text.strip(), 'headline': a.text})

df = pd.DataFrame(rows)
print(len(df), 'عنوان')
df.head(10)
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

companies = {'AAPL': 'Apple', 'TSLA': 'Tesla', 'NVDA': 'Nvidia'}
headers = {'User-Agent': 'Mozilla/5.0'}
rows = []

for ticker, name in companies.items():
    url = f'https://news.google.com/rss/search?q={name}+stock&hl=en-US&gl=US&ceid=US:en'
    r = requests.get(url, headers=headers, timeout=15)
    soup = BeautifulSoup(r.text, 'xml')
    for item in soup.find_all('item'):
        rows.append({
            'ticker': ticker,
            'date': item.pubDate.text,
            'headline': item.title.text,
        })

df = pd.DataFrame(rows)
print(len(df), 'عنوان')
df.head(10)
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer

sia = SentimentIntensityAnalyzer()

df['score'] = df['headline'].apply(lambda h: sia.polarity_scores(h)['compound'])
df['label'] = df['score'].apply(
    lambda s: 'positive' if s >= 0.05 else ('negative' if s <= -0.05 else 'neutral')
)

print(df.groupby('ticker')['score'].agg(['count', 'mean']).round(3))
df[['ticker', 'headline', 'score', 'label']].head(10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# إضافة كلمات مالية لقاموس VADER
sia.lexicon.update({
    'record': 2.0, 'high': 1.5, 'rally': 2.5, 'surge': 2.5, 'soar': 2.5,
    'beat': 2.0, 'beats': 2.0, 'upgrade': 2.0, 'gain': 1.8, 'gains': 1.8,
    'falls': -2.0, 'fall': -2.0, 'drop': -2.0, 'plunge': -3.0, 'slump': -2.5,
    'sell-off': -2.5, 'selloff': -2.5, 'downgrade': -2.0, 'bearish': -2.0,
    'bullish': 2.0, 'miss': -2.0, 'misses': -2.0, 'lawsuit': -2.0, 'trillion': 0,
})

df['score'] = df['headline'].apply(lambda h: sia.polarity_scores(h)['compound'])
df['label'] = df['score'].apply(
    lambda s: 'positive' if s >= 0.05 else ('negative' if s <= -0.05 else 'neutral')
)

sns.set_theme(style='whitegrid')
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.barplot(data=df, x='ticker', y='score', errorbar=None, ax=axes[0])
axes[0].set_title('Average Sentiment Score')
axes[0].axhline(0, color='black', linewidth=0.8)

counts = df.groupby(['ticker', 'label']).size().unstack(fill_value=0)
counts.plot(kind='bar', stacked=True, ax=axes[1],
            color={'negative': '#d9534f', 'neutral': '#bbbbbb', 'positive': '#5cb85c'})
axes[1].set_title('Headlines by Sentiment')
axes[1].tick_params(axis='x', rotation=0)

plt.tight_layout()
plt.show()
```

```python
df.to_csv('financial_news_sentiment.csv', index=False)
```

```python
from transformers import pipeline

finbert = pipeline('text-classification', model='ProsusAI/finbert', top_k=None)

def finbert_score(headline):
    res = finbert(headline[:512])[0]
    d = {r['label']: r['score'] for r in res}
    return d['positive'] - d['negative']   # من -1 إلى +1

print(finbert_score('Apple Stock Hits Record High On iPhone Duo Launch'))
print(finbert_score('Apple Stock Falls Today as Bearish Analyst Comments'))
```

```python
import requests, nltk
from bs4 import BeautifulSoup
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from nltk.sentiment.vader import SentimentIntensityAnalyzer

nltk.download('vader_lexicon', quiet=True)

# 1. سحب الأخبار
companies = {'AAPL': 'Apple', 'TSLA': 'Tesla', 'NVDA': 'Nvidia'}
headers = {'User-Agent': 'Mozilla/5.0'}
rows = []
for ticker, name in companies.items():
    url = f'https://news.google.com/rss/search?q={name}+stock&hl=en-US&gl=US&ceid=US:en'
    soup = BeautifulSoup(requests.get(url, headers=headers, timeout=15).text, 'xml')
    for item in soup.find_all('item'):
        rows.append({'ticker': ticker, 'date': item.pubDate.text, 'headline': item.title.text})
df = pd.DataFrame(rows)

# 2. تحليل المشاعر مع القاموس المالي
sia = SentimentIntensityAnalyzer()
sia.lexicon.update({
    'record': 2.0, 'high': 1.5, 'rally': 2.5, 'surge': 2.5, 'soar': 2.5,
    'beat': 2.0, 'beats': 2.0, 'upgrade': 2.0, 'gain': 1.8, 'gains': 1.8,
    'falls': -2.0, 'fall': -2.0, 'drop': -2.0, 'plunge': -3.0, 'slump': -2.5,
    'sell-off': -2.5, 'selloff': -2.5, 'downgrade': -2.0, 'bearish': -2.0,
    'bullish': 2.0, 'miss': -2.0, 'misses': -2.0, 'lawsuit': -2.0, 'trillion': 0,
})
df['score'] = df['headline'].apply(lambda h: sia.polarity_scores(h)['compound'])
df['label'] = df['score'].apply(
    lambda s: 'positive' if s >= 0.05 else ('negative' if s <= -0.05 else 'neutral'))

# 3. حفظ البيانات
df.to_csv('financial_news_sentiment.csv', index=False)

# 4. الرسم والحفظ
sns.set_theme(style='whitegrid')
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
sns.barplot(data=df, x='ticker', y='score', errorbar=None, ax=axes[0])
axes[0].set_title('Average Sentiment Score')
axes[0].axhline(0, color='black', linewidth=0.8)
counts = df.groupby(['ticker', 'label']).size().unstack(fill_value=0)
counts.plot(kind='bar', stacked=True, ax=axes[1],
            color={'negative': '#d9534f', 'neutral': '#bbbbbb', 'positive': '#5cb85c'})
axes[1].set_title('Headlines by Sentiment')
axes[1].tick_params(axis='x', rotation=0)
plt.tight_layout()
plt.savefig('sentiment_overview.png', dpi=150)
plt.show()

print(len(df), 'headlines saved')
```
